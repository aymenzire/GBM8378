<div> a) Lorsque la fréquence est plus élevée, les ondes ultrasonores perdent plus rapidement leur énergie en traversant les tissus, car elles sont davantage atténuées et absorbées. Ca fait que le signal devient trop faible après une certaine distance. À l’inverse, une fréquence plus basse est moins absorbée, ce qui permet aux ultrasons de pénétrer plus profondément dans le corps, mais avec moins de détails dans l’image.  
    
b) Lorsque la fréquence est plus élevée, la longueur d’onde des ultrasons devient plus petite. Cela permet au système de distinguer des structures plus petites et plus rapprochées, ce qui améliore la résolution, surtout dans la direction de propagation du faisceau. Une fréquence plus élevée peut aussi améliorer la résolution latérale, car lorsque la fréquence augmente, la longueur d’onde diminue, ce qui rend le faisceau plus étroit. (Resolution latereale = F x (lambda)/D). À l’inverse, une fréquence plus basse donne une image moins détaillée, car la longueur d’onde est plus grande et les structures proches peuvent être plus difficiles à distinguer.

c) Dans les tissus biologiques, l’absorption augmente presque linéairement avec la fréquence, car b environ 1 dans tissus biologiques. Ca signifie que plus la fréquence est élevée, plus l’énergie de l’onde ultrasonore est transformée en chaleur dans le tissu, ce qui diminue  l’amplitude du signal pendant sa propagation. Cette relation est décrite par l’expression suivante :
α = 𝑎𝑓^(𝑏) 
Cela montre que lorsque la fréquence augmente, le coefficient d’atténuation augmente aussi.
</div>

<div class="alert alert-block alert-warning"><b>1.2. Quelle formule permet de calculer le temps $t$ après lequel l’écho d'un signal émis à $t_0=0s$ sera detecté par le transducteur ? (1pt)</b></div>

<div> La formule vient du fait que l’onde ultrasonore fait un aller-retour entre le transducteur et l’interface du tissu. Quand le transducteur émet une onde à t0=0, l’onde se propage dans le tissu jusqu’à une interface située à une distance d. À cette interface, une partie de l’onde est réfléchie et revient vers le transducteur. Donc l’onde parcourt deux fois la distance d (Distance totale = 2d). On utilise ensuite la relation entre distance, vitesse et temps : distance = vitesse × temps. Donc, t= 2d/c où t= temps pour recevoir l’écho, d = profondeur de l’interface, c = vitesse du son dans le tissu.
</div>

<div> Premièrement, elles peuvent subir une réflexion, lorsqu’elles rencontrent une interface entre deux milieux différents et rebondissent, comme un écho. <div class="alert alert-block alert-warning"><b>1.6. Expliquer pourquoi il n’y a pas de problème à faire transiter le signal d’aller et de retour par le même canal du transducteur au pulseur. (1pt)</b></div>

<div> Il n’y a pas de problème parce que l’émission et la réception ne se font pas simultanément. Le transducteur émet d’abord une impulsion, puis il passe en mode réception pour capter les échos (soit un seul à la fois). Donc le signal d’aller et le signal de retour utilisent le même canal, mais à des instants différents. <div class="alert alert-block alert-warning"><b>1.7. Prouver que pour un signal périodique, de la forme $f(t)= sin (\omega_0t)$, la transformée de Fourier fait ressortir le contenu fréquentiel. (1pt)</b></div>

<div > Pour le signal périodique ( f(t)=sin(w_0 t) ), on commence par l’écrire sous forme exponentielle :

[sin(w_0*t)=(1/2i)(e^[iw_0*t]-e^[-iw_0*t])]

En utilisant la linéarité de la transformée de Fourier et le fait que : 
    TF[e^(iw_0*t)]= delta(w-w_0) 
    et 
    TF[e^(-iw_0*t)]=delta(w+w_0)

on obtient : TF{sin(w_0*t)}=(1/2i)(delta(w-w_0)-delta(w+w_0))

Donc, la transformée de Fourier est constituée de deux impulsions de Dirac situées aux fréquences w=+/-w_0, ce qui montre que le signal ne contient qu’une seule fréquence. Démontrant que la transformée de Fourier met directement en évidence le contenu fréquentiel du signal. </div>

<div class="alert alert-block alert-warning"><b>2.1. Le signal <I>scanA.npy</I> représente la réponse ultrasonore du scan de deux tiges identiques, localisées à différentes profondeurs. Interprétez la décroissance du signal. (0.5pt)</b></div>

```python
# On charge le fichier'scanA.npy' et on l'affiche. 
scanA = np.load('scanA.npy')

plt.figure(figsize=[8, 8])
plt.plot(scanA)
plt.title("Signal ScanA")
plt.xlabel("Echantillon n$\degree$")
plt.ylabel("Amplitude de l'onde sonore");
```

<div>La décroissance du signal observée dans le scan A s’explique principalement par deux phénomènes  : l’atténuation exponentielle et la divergence géométrique de l’onde. En effet, l’amplitude des ultrasons suit une loi de type A(z) = A<sub>0</sub> e<sup>-μz</sup> , ce qui traduit une perte d’énergie progressive due à l’absorption du milieu (conversion en chaleur). À cela s’ajoute un facteur géométrique en 1/r, lié à l’expansion sphérique de l’onde, qui diminue aussi l’amplitude avec la distance .
On observe que la décroissance est relativement lente et régulière, ce qui est cohérent avec les tissus mous où le coefficient d’atténuation est faible (seulement ~ 1 % réfléchi aux interfaces). Cela donne une enveloppe quasi exponentielle décroissante plutôt douce.
Les gros pics visibles sur le signal correspondent aux réflexions aux interfaces des deux tiges. Ces pics sont beaucoup plus élevés que le signal de fond car ils résultent d’une forte différence d’impédance acoustique (Z) entre les tiges et le milieu environnant, ce qui augmente fortement le coefficient de réflexion.
Enfin, le deuxième pic est plus faible que le premier, car il provient d’une plus grande profondeur : l’onde a subi davantage d’atténuation (terme exponentiel) et de dispersion (1/r), ce qui réduit son amplitude avant d’être détectée. </div>

<div class="alert alert-block alert-warning">
<b>2.2. Proposez une méthode pour corriger cette décroissance et obtenir un signal de fond quasi-nul. Appliquez cette méthode puis affichez l’écho corrigé obtenu. (1pt)</b>
    
**Astuce:** Trouvez un approximation de la courbe de décroissance pour pouvoir la compenser par la suite.
</div>

<div >

Pour corriger la décroissance du signal de fond, on peut approximer l’enveloppe lente du signal par une fonction de tendance, puis la soustraire au signal original. Comme l’énoncé du TP rappelle que l’amplitude décroît avec la profondeur à cause de l’atténuation et de la divergence géométrique, une approximation exponentielle serait physiquement justifiée (voir question 2.1). Cependant, pour un traitement simple et robuste dans ce signal, on peut aussi utiliser un ajustement polynomiale de faible ordre sur le fond, en excluant l’effet rapide des grands pics. Cette tendance représente alors le fond décroissant du signal.
En soustrayant cette courbe ajustée au signal initial, on obtient un signal corrigé dont le fond devient quasi nul, tandis que les échos des deux tiges restent bien visibles sous forme de pics marqués.

</div>

```python
plt.figure(figsize=[8, 8])

# Commencez votre code ici (7 lignes max)
x = np.arange(len(scanA))
p = np.polyfit(x, scanA, 3)
fond = np.polyval(p, x)
echo_corrige = scanA - fond

plt.plot(echo_corrige)
plt.title("Echo corrigé")
plt.xlabel("Echantillon n$\degree$")
plt.ylabel("Amplitude de l'onde sonore");
```

<div class="alert alert-block alert-warning"><b>2.3. Sachant que la fréquence d’échantillonnage est de 100 MHz et que la vitesse du son dans l'eau est de 1500 m/s, à quelle distance est placée chacune des tiges? (0.5pt) </b></div>

```python
Fs = 100e6  # Fréquence d'échantillonage (Hz)
Vs = 1500  # Célérité du son dans l'eau (m/s)

# Commencez votre code ici (7 lignes max)

n1 = 1000
n2 = 1500
# Conversion échantillon → temps : t = n / Fs (car chaque échantillon correspond à 1/Fs secondes)
t1 = n1 / Fs
t2 = n2 / Fs

# Relation échographie : d = v * t / 2 (le facteur 1/2 vient du fait que l'onde fait un aller-retour)
d1 = Vs * t1 / 2
d2 = Vs * t2 / 2

print(f"La distance de la première tige est de {d1:.5f} m.")
print(f"La distance de la deuxième tige est de {d2} m.")
```

<div class="alert alert-block alert-warning"><b>2.4. Effectuez la transformée de Fourier numérique des données recueillies afin de retrouver la réponse en fréquence du transducteur. Affichez le résultat obtenu ainsi que les données fournies par le constructeur (<I>H.npy</I>) sur un même graphique, identifiez les axes et la fréquence centrale. (1pt)</b></div>

```python

plt.figure(figsize=[8, 8])

# Commencez votre code ici (3 lignes max)
S = np.abs(fftshift(fft(echo_corrige, 2048))); f = np.linspace(-Fs/2, Fs/2, 2048)/1e6
plt.plot(f, S/np.max(S), 'b--', linewidth=2, label='TF des données')
fc = f[1024 + np.argmax(S[1024:])]

# Comparer avec les données constructeur
H = np.load('H.npy')
# Interpolate between 0 and 50 Mhz
nbpts = 2048 # Nombre de données acquises par 1 transducteur
H = np.interp(np.linspace(0, 50, nbpts // 2 + 1), np.linspace(0, 30, len(H)), H, left=0, right=0)
H = np.concatenate((H[-1:0:-1], H[:-1]))


plt.plot(np.linspace(-Fs/2,Fs/2,2048)/1e6, H/np.max(H), color='red', linewidth=2, label='Constructeur H.npy')
plt.xlabel("Fréquence (MHz)")
plt.ylabel("Amplitude normalisée")
plt.title(f"Réponse en fréquence du transducteur, f_c = {fc:.2f} MHz")
plt.legend();
```

<div class="alert alert-block alert-warning">
<b>3.1. Comment interprétez-vous cette convolution? Pourquoi le signal recueilli diffère-t-il du signal réel ? </b><br>
    La réponse du transducteur est fournie par le fournisseur dans le fichier H.npy (1pt)
</div>

<div> Le signal que le transducteur mesure n’est jamais exactement le signal réel, parce que le signal réel représente ce qui se passe vraiment dans le fantôme, comme la position exacte des billes et les réflexions acoustiques qu’elles provoquent.Le transducteur ne réagit pas de manière uniforme à toutes les fréquences : certaines sont amplifiées, d’autres atténuées. Ces caractéristiques sont décrites par H.npy. Chaque valeur de H.npy montre comment le transducteur modifie le signal à un moment donné ou pour une certaine fréquence. La convolution décrit comment chaque point du signal réel est combiné avec H.npy pour donner signal final mesuré.

Le signal recueilli diffère du signal réel pour deux raisons principales : d’abord, le transducteur n’est pas parfait et a sa propre réponse (H.npy); ensuite, le signal réel est spatial, mais la mesure est faite dans le temps, donc ce que le transducteur voit est une version “modifiée et étalée” du signal réel.</div>

<div class="alert alert-block alert-warning"><b>
3.2. Résoudre l’équation (1) pour obtenir le signal réel f(t) sachant la réponse du transducteur g(t). (1pt)</b></div>

<div> Si on veut trouver le signal réel f(t) à partir du signal mesuré g(t) et de la réponse du transducteur h(t), on utilise le principe inverse de la convolution. Donc, on utilise la déconvolution. On passe g(t) et h(t) en domaine fréquence avec Fourier. On divise G(f) par H(f) point par point. Puis, on revient dans le domaine temps avec la transformée inverse.
f(t)= TF^−1 [ TF[g(t)] / TF[h(t)] ] </div>

<div class="alert alert-block alert-warning"><b>3.3. En pratique un terme de régularisation µ est ajouté et l’équation utilisée est : <br>
    $F(\omega)=\frac{G(\omega)\cdot H(\omega)}{H(\omega)^2+\mu}$<br>
    
Quelle est l’utilité de ce terme de régularisation ? (1pt)<br>

</b>
</div>

<div> Le terme de régularisation μ sert à stabiliser la déconvolution lorsque certaines fréquences de (H(w)) sont très faibles ou proches de zéro. Sans ce terme, diviser par (H(w)) dans ces zones amplifie énormément le bruit présent dans le signal mesuré  (G(w)), ce qui donnerait un signal réel (F(w)) très bruité. En ajoutant (μ) dans le dénominateur, on empêche cette amplification excessive, même si cela introduit une petite erreur sur le signal réel.En pratique, (μ) est choisi pour trouver un compromis entre  fidélité au signal original et limitation du bruit, pour obtenir une estimation plus stable et fiable de (f(t)).</div>

<div class="alert alert-block alert-warning"><b>3.4. Effectuez la déconvolution de l’image <I>rawdata.npy</I> (en utilisant <I>H.npy</I>) et comparer avec les données de <I>rawdata.npy</I>. Vous pouvez utiliser $\mu=5\times10^{-8}$. (2pts)</b></div>

```python
# Création d'un subplot 1 par 2
plt.figure(figsize=[10, 5])
plt.subplot(1,2,1)

# Chargement et plot de rawdata
rawdata = np.load('rawdata.npy')
plt.imshow(rawdata, aspect='auto', vmin=-1, vmax=1)
plt.title("rawdata")
plt.colorbar();

# Commencez votre code ici

mu = 5e-8
N = rawdata.shape[0]

# Chargement de H
Hsmall = np.load('H.npy').squeeze()

# H avec les même dimensions compatible avec G
H = np.zeros((N,1), dtype=complex)
H[:len(Hsmall), 0] = Hsmall

# signal non shifté à fourier non shifté avec fft 
G = fft(rawdata, axis=0)


# Déconvolution régularisée avec conj(H) avec la formule de la question 3.3
F = G * np.conj(H) / (np.abs(H)**2 + mu)

# Retour en domaine spatial avec ifft
deconv = np.real(ifft(F, axis=0))



plt.subplot(1,2,2)
plt.imshow(deconv, aspect='auto', vmin=-1, vmax=1)
plt.title("Signal non shifté après déconvolution")
plt.colorbar();
plt.tight_layout()
```

<div class="alert alert-block alert-warning"><b>3.5. Que se passe-t-il au niveau des bords supérieurs et inférieurs ? (1pt)</b></div>

Aux bords supérieur et inférieur, on observe des artefacts importants, tels que des zones saturées noir et des lignes épaisses. Cela s’explique par le fait que la transformée de Fourier suppose que le signal est périodique, ce qui introduit des discontinuités artificielles aux extrémités de l’image. On observe que l'image est compressée verticalement sur la transformé de fourier de l'image à droite ce qui explique aussi les artefacts. Lors de la déconvolution, ces erreurs sont amplifiées, rendant les distorsions encore plus visibles aux bords.
</div>


<div class="alert alert-block alert-warning"><b>3.6. À quoi correspondent les barres horizontales visibles sur une partie de l’image ? (1pt)</b></div>

<div> Les barres horizontales observées dans l’image correspondent à la réponse impulsionnelle du transducteur. Après déconvolution, ces structures deviennent plus visibles car le processus amplifie certaines composantes du signal, révélant ainsi ces répétitions liées à la réponse du système.
```python
## Title: Ultrasound Forward Simulator
## Filename: __main__.py
## Authors: Jonathan Poree, Samuel Desmarais --- Provost Ultrasound Lab
## Inspired from work in: Section 2.3 of https://doi.org/10.1088/1361-6560/aae3c3

from labUSfunctions import ussimforward, prepare_animation

##################
### PARAMETERS ###
##################

## PROBE
prm = {}
prm['fc']       = 15.625e6  # Central frequency of the transducer (Hz)
prm['pitch']    = 0.1e-3  # Spacing between the center of 2 elements (m)
prm['width']    = 0.8117*1540/prm['fc']  # Width of element (m)
prm['height']   = 1.5e-3  # Height of element (m)
prm['nele']     = 1  # Number of elements
prm['BW']       = 0.5  # 50% Pulse Bandwidth
prm['eleBW']    = 1  # 100% Bandwidth
prm['hardbaffle'] = True  # Use hard baffle =True or soft baffle =False

## MEDIUM
prm['c']        = 1540  # Propagation velocity (m/s)
prm['lbd']      = prm['c']/prm['fc']  # Wavelenght (m)

## IMAGING SEQUENCE
prm['fs']        = 4 * prm['fc']  # Sampling frequency of the system (Hz)
prm['elevfocus'] = 64 * prm['lbd']
prm['fspulse']   = 10*prm['fs']  # Sampling frequency for the pulse (Hz)
prm['pulse']     = np.sin(2*np.pi*prm['fc']*np.arange(0, 2/prm['BW']/prm['fc']+1/prm['fspulse'], 1/prm['fspulse']))
prm['pulse']     = prm['pulse'] * np.hanning(prm['pulse'].size)  # Shape of the pulse
prm['planewaveangle'] = 0  # Plane wave emission angle (degrees)

## SIMULATION
prm['useprobeheight'] = False  # Take probe height into account =True or not =False
prm['gridsize'] = 128  # Number of samples per side of the grid
prm['fovx'] = prm['lbd'] * 64  # Lateral field of view (m)
prm['fovz'] = prm['lbd'] * 128  # Axial field of view (m)
prm['floatprecision'] = np.float32  # Float arrays precision
prm['complexprecision'] = np.complex64  # Complex arrays precision

#############
### SETUP ###
#############

## Grid Definition
zgrid, xgrid = np.meshgrid(np.linspace(0, 1, prm['gridsize'])*prm['fovz'], np.linspace(-0.5, 0.5, prm['gridsize'])*prm['fovx'])
ygrid = np.zeros_like(xgrid)

## Delays Definition
delays = np.array([0.])

## Apodization Definition
apod = np.ones_like(delays)

##################
### SIMULATION ###
##################
RF = ussimforward(xgrid, ygrid, zgrid, delays, apod, prm)  # Call to the simulation function

######################
### OUTPUT US WAVE ###
######################
anim = prepare_animation(RF, prm['fovx'], prm['fovz'])
display(HTML(anim.to_html5_video()))
```

Le champ de pression simulé montre la propagation d’une onde ultrasonore émise par un transducteur composé d'un seul élément (monoélément) au cours du temps. On observe que l’onde se propage axialement (selon l’axe z) à partir du transducteur situé en haut de l’image, ce qui est cohérent avec une émission normale (angle nul). La position de l’onde se déplace progressivement vers des profondeurs plus grandes, illustrant la propagation à vitesse constante dans le milieu. La forme du front d’onde est courbée (quasi circulaire), ce qui indique une propagation sphérique. Cela s’explique par le fait que le transducteur possède une ouverture finie et n’est pas focalisé : l’énergie se disperse dans l’espace, ce qui correspond au phénomène de diffraction. En conséquence, le faisceau s’élargit avec la profondeur, ce qui est également visible dans l’image intégrée à droite.

L’image de droite (intégration temporelle) met en évidence l’enveloppe du champ de propagation. On y observe une intensité maximale le long de l’axe central, avec une décroissance latérale et axiale. Cela montre que l’énergie acoustique est concentrée autour de l’axe du transducteur mais diminue en profondeur à cause de la divergence géométrique du faisceau et de la répartition spatiale de l’énergie.

<div class="alert alert-block alert-warning"><b>4.2. Quel type de propagation est une onde sonore ? Écrivez la forme mathématique générale de ce type d’onde. (1pt) </b></div>

<div> Une onde sonore est une onde mécanique longitudinale qui se propage par des variations de pression dans le milieu, les particules oscillant dans la même direction que la propagation.

Sa forme mathématique générale est donnée par :
p(r,t) = (A / r) f(t − r / c),

où A est l’amplitude, r la distance, c la vitesse du son et f la forme temporelle de l’onde. Cette expression décrit une onde progressive se propageant dans le milieu avec une atténuation géométrique en 1/r.
</div>

<div class="alert alert-block alert-warning"><b>4.5. Toujours avec 10 éléments, faire une fonction qui viendra modifier le vecteur 'delays' afin d’obtenir une onde plane à un angle de 10°. Affichez le front d’onde à 1.5µs. (1.5pts) </b><br>

```python

##################
### PARAMETERS ###
##################

## PROBE
prm = {}
prm['fc']       = 15.625e6
prm['pitch']    = 0.1e-3
prm['width']    = 0.8117*1540/prm['fc']
prm['height']   = 1.5e-3
prm['nele']     = 10
prm['BW']       = 0.5
prm['eleBW']    = 1
prm['hardbaffle'] = True

## MEDIUM
prm['c']        = 1540
prm['lbd']      = prm['c']/prm['fc']

## IMAGING SEQUENCE
prm['fs']        = 4 * prm['fc']
prm['elevfocus'] = 64 * prm['lbd']
prm['fspulse']   = 10 * prm['fs']
prm['pulse']     = np.sin(2*np.pi*prm['fc']*np.arange(0, 2/prm['BW']/prm['fc'] + 1/prm['fspulse'], 1/prm['fspulse']))
prm['pulse']     = prm['pulse'] * np.hanning(prm['pulse'].size)
prm['planewaveangle'] = 10

## SIMULATION
prm['useprobeheight'] = False
prm['gridsize'] = 128
prm['fovx'] = prm['lbd'] * 64
prm['fovz'] = prm['lbd'] * 128
prm['floatprecision'] = np.float32
prm['complexprecision'] = np.complex64

#############
### SETUP ###
#############

zgrid, xgrid = np.meshgrid(
    np.linspace(0, 1, prm['gridsize']) * prm['fovz'],
    np.linspace(-0.5, 0.5, prm['gridsize']) * prm['fovx']
)
ygrid = np.zeros_like(xgrid)

#ajouter la fonction pour ajouter le délai

def plane_wave_delays(nele, pitch, angle_deg, c):
    theta = np.deg2rad(angle_deg)
    xele = (np.arange(nele) - (nele - 1)/2) * pitch
    delays = xele * np.sin(theta) / c
    return delays - delays.min()   # pour avoir des délais positifs


delays = plane_wave_delays(prm['nele'], prm['pitch'], 10, prm['c'])
apod = np.ones(prm['nele'])

##################
### SIMULATION ###
##################
RF = ussimforward(xgrid, ygrid, zgrid, delays, apod, prm)
anim = prepare_animation(RF, prm['fovx'], prm['fovz'])
display(HTML(anim.to_html5_video()))

#############################
### FRONT D'ONDE À 1.5 µs ###
#############################
t = 1.5e-6
idx = int(round(t * prm['fs']))

plt.figure(figsize=[6, 8])
plt.imshow(
    RF[:, :, idx],
    extent=[-prm['fovx']/2*1e3, prm['fovx']/2*1e3, prm['fovz']*1e3, 0],
    cmap='seismic',
    aspect='auto'
)
plt.xlabel("Distance (mm)")
plt.ylabel("Distance (mm)")
plt.title(f"Champ de pression à t = {t*1e6:.1f} µs, indice = {idx}")
plt.colorbar(label="Pression");
```

<div class="alert alert-block alert-warning"><b>4.6. Quel est l’intérêt de pouvoir focaliser en émission? Y a-t-il un désavantage à utiliser un grand nombre de transducteurs répartis sur une large distance? (1pt)</b>
</div>

<div> 
La focalisation en émission permet de concentrer l’énergie ( énergie acoustique) à une profondeur donnée, ce qui augmente l’intensité du signal dans la zone d’intérêt et améliore la résolution spatiale, en particulier la résolution latérale, en raison des interférences constructives et déstructive qui interviennent lors de la focalisation de l'onde. Cela permet d’obtenir des images plus précises et de mieux distinguer des structures proches.
Utiliser un grand nombre de transducteurs répartis sur une large distance peut présenter des désavantages. Bien que cela améliore la directivité et la focalisation du faisceau, cela peut entraîner l’apparition de lobes secondaires plus marqués dus aux interférences. Cela peut dégrader la qualité de l’image. De plus, comme désavantage secondaire cela augmente la complexité du système (contrôle des délais, électronique) et peut limiter la zone de focalisation à une profondeur spécifique.
</div>

<div class="alert alert-block alert-warning"><b>4.8. Toujours avec 10 élements, adaptez le paramètre 'delays' afin d'effectuer un focus en x=0.5mm et z=2mm. Expliquez vos résultats. (1.5pts)</b>
</div>

```python
##################
### PARAMETERS ###
##################

## PROBE
prm = {}
prm['fc']       = 15.625e6  # Central frequency of the transducer (Hz)
prm['pitch']    = 0.1e-3  # Spacing between the center of 2 elements (m)
prm['width']    = 0.8117*1540/prm['fc']  # Width of element (m)
prm['height']   = 1.5e-3  # Height of element (m)
prm['nele']     = 10  # Number of elements
prm['BW']       = 0.5  # 50% Pulse Bandwidth
prm['eleBW']    = 1  # 100% Bandwidth
prm['hardbaffle'] = True  # Use hard baffle =True or soft baffle =False

## MEDIUM
prm['c']        = 1540  # Propagation velocity (m/s)
prm['lbd']      = prm['c']/prm['fc']  # Wavelenght (m)

## IMAGING SEQUENCE
prm['fs']        = 4 * prm['fc']  # Sampling frequency of the system (Hz)
prm['elevfocus'] = 64 * prm['lbd']
prm['fspulse']   = 10*prm['fs']  # Sampling frequency for the pulse (Hz)
prm['pulse']     = np.sin(2*np.pi*prm['fc']*np.arange(0, 2/prm['BW']/prm['fc']+1/prm['fspulse'], 1/prm['fspulse']))
prm['pulse']     = prm['pulse'] * np.hanning(prm['pulse'].size)  # Shape of the pulse
prm['planewaveangle'] = 0  # Plane wave emission angle (degrees)

## SIMULATION
prm['useprobeheight'] = False  # Take probe height into account =True or not =False
prm['gridsize'] = 128  # Number of samples per side of the grid
prm['fovx'] = prm['lbd'] * 32  # Lateral field of view (m)
prm['fovz'] = prm['lbd'] * 64  # Axial field of view (m)
prm['floatprecision'] = np.float32  # Float arrays precision
prm['complexprecision'] = np.complex64  # Complex arrays precision

#############
### SETUP ###
#############

## Grid Definition
zgrid, xgrid = np.meshgrid(np.linspace(0, 1, prm['gridsize'])*prm['fovz'], np.linspace(-0.5, 0.5, prm['gridsize'])*prm['fovx'])
ygrid = np.zeros_like(xgrid)

#fonction utilisé pour faire le focus en utilisant des délai

def focus_delays(nele, pitch, xf, zf, c):
    xele = (np.arange(nele) - (nele - 1)/2) * pitch
    r = np.sqrt((xele - xf)**2 + zf**2)
    delays = (r.max() - r) / c
    return delays

## Delays Definition
delays = focus_delays(prm['nele'], prm['pitch'], 0.5e-3, 2e-3, prm['c'])

## Apodization Definition
apod = np.ones_like(delays)

##################
### SIMULATION ###
##################
RF = ussimforward(xgrid, ygrid, zgrid, delays, apod, prm)  # Call to the simulation function

######################
### OUTPUT US WAVE ###
######################
anim = prepare_animation(RF, prm['fovx'], prm['fovz'])
display(HTML(anim.to_html5_video()))
```

<div> 
Pour focaliser en émission au point x = 0.5 mm et z = 2 mm, on applique à chaque élément un délai dépendant de sa distance au point focal. Les éléments les plus éloignés de ce point émettent en premier, tandis que les plus proches émettent plus tard, de sorte que toutes les ondes arrivent en phase au foyer.
On observe alors que l’énergie acoustique se concentre autour du point visé. Cela forme un faisceau convergent avant le foyer puis divergent après celui ci. L’intensité du champ devient maximale près de la position demandée, soit autour de x = 0.5 mm et z = 2 mm. Cela montre que les délais permettent de contrôler spatialement la zone de focalisation et d’améliorer la résolution latérale dans la région d’intérêt.
</div>
